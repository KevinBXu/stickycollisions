```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# parameters

# [KRb+Rb, RbCs+RbCs]
temperatures = [480e-9, 2.2e-6] # K
mass = [51, 110] # amu
vdW = [7700, 1.9e5] # a.u.
rotational_constant = [1.7e-7, 7.7e-8] # a.u.

species = 0

e_mass = 5.48579909e-4    # electron mass in amu
t0 = 315775               # a.u. to K
tau0 = 2.418884326509e-17 # a.u. to sec\

mu = mass[species] / e_mass        
c6 = vdW[species]             
brot = rotational_constant[species]   
temp = temperatures[species] 

beta = (2 * mu * c6) ** (1/4)    # length 
E_beta = 1 / (2 * mu * beta ** 2)   # energy
tau_beta = 1 / E_beta    # time
print("reduced mass (a.u.)", mu)
print("C6 (au)", c6)
print("C6 (E_beta * beta**6)", c6 / E_beta / beta**6)
print("beta (a0)", beta)
print("E_beta (K)", E_beta * t0)
print("tau_beta (s)", tau_beta * tau0)
```

```python
# parameters
r0 = (c6 / brot) ** (1/6) # a.u.
r1 = beta # a.u.
print("r0", r0)
print("r1", r1)
print("E0", c6/r0**6)
print("E1", c6/r1**6)
dE = (c6/r0**6 - c6/r1**6) / E_beta
print("dE (au)", dE * E_beta)
print("dE (K)", dE * E_beta * t0)
print("dE (E_beta)", dE)
print("dr (beta)", 1 - r0 / beta)
```

```python
kbT = temp / t0 / E_beta
print("kbT (E_beta)", kbT)
q = np.sqrt((kbT+dE))
k = np.sqrt((kbT))

R = (q-k)**2 / (q+k)**2
T = k/q * 4*q**2 / (q+k)**2
print("r", R)
print("t", T)
```
